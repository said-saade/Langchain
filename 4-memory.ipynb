```python
##ChatOpenAI(): cria o modelo de linguagem (por exemplo, GPT-4 via API da OpenAI).
##ConversationChain: conecta o modelo com a memória para manter contexto entre interações.
## verbose=True: mostra detalhes do processamento.

from langchain_openai.chat_models import ChatOpenAI
from langchain.chains.conversation.base import ConversationChain

chat = ChatOpenAI()
memory = ConversationBufferMemory()
conversation = ConversationChain(
    llm=chat,
    memory=memory,
    verbose=True
)
```

```python
## Aqui a 1° pergunta feita, que sera gerada uma resposta
conversation.predict(input="Olá, meu nome é Rodrigo")
```

```python
## Aqui a 2° pergunta que levara em consideração a resposta da 1° pergunta
## demonstrando que a utilização da memoria esta funcionando.
conversation.predict(input="Como vai?")
```
